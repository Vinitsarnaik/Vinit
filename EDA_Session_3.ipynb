### **Pie Chart**

- autopct = %a.bf%%

- first % is the formatting

- a : how much min width

- b : after point how many decimals

- %% : % symbol

- %1.1f%% : 40.2%

- %2.2f%% : 423.23%

```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
```

```python
plt.pie(values,
        labels=keys,
       autopct='%1.2f%%',
       startangle=40,
        explode=[0.05 for i in range (6)],
        radius=0.9
       )
plt.savefig('continent_pie.jpg')
plt.show()
```

```python
# plt.pie(x,labels)
colors=['Red','Blue','sandybrown','Yellow','Pink','Purple']
plt.pie(values,
       labels=keys,
       colors=colors,
       radius=0.8,     # fro changing the size of the pie chart
       explode=[0.1 for i in range(6)],  # for seperating the slices of pie chart by 0.1 unit
       autopct='%1.2f%%',    # for showing the percentage of the slice
       startangle=40)   # for changing the angle and rotation of the pie chart
plt.savefig('Continent_pie.jpg')

plt.show()
```
