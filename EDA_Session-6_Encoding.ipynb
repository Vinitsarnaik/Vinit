**pd.get_dummies**

- get dummies is a way to represent the categorical columns to numerical 

- first it will check what are the labels in a column

- for example case status has two labels 

    - denied and certified

- So it will create two new columns based on labels

    - col1: case_status_Denied
    - col2: case_status_Certified


|case_status| case_status_Denied | case_status_Certified |
|    -      |         -          |      -                |
|Denied     |         1          |      0                |
|Certified  |         0          |      1                |
|Certified  |         0          |      1                |

**It is also called as one Hot Encode**

- It makes new columns are independent each other

- Some of ML algorithms except that columns should be **Independent** each other

```python
pd.get_dummies(visa_df['case_status'],
               dtype=int)
```
